20. Plot Continuous Variables

```python
cate_val= []
cont_val= []
```

```python
for column in data.columns:
  if data[column].nunique() <=10:
    cate_val.append(column)
  else:
    cont_val.append(column)
```

```python
data.hist(column=cont_val, figsize=(10,10), layout=(3,2))
plt.tight_layout()
plt.show()
```
